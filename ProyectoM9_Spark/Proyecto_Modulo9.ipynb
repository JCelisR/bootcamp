# Proyecto Módulo 9: Pipeline de Big Data - RetailMax

### Descripción del Proyecto
Este proyecto implementa un flujo completo de procesamiento de datos para la empresa "RetailMax". El objetivo es transformar datos brutos de ventas en información estratégica mediante el uso de arquitecturas de Big Data, aplicando limpieza, análisis estadístico (SQL) y modelos de aprendizaje automático (Machine Learning).

### Objetivos de Aprendizaje
* **Ingesta y RDDs:** Manipulación de colecciones de datos distribuidos.
* **Spark SQL:** Procesamiento de datos estructurados mediante consultas de agregación.
* **Machine Learning:** Segmentación de clientes utilizando el algoritmo de clustering K-Means.

---

```python
import pandas as pd
from sklearn.cluster import KMeans
import numpy as np

print("--- 🚀 EJECUTANDO PIPELINE RETAILMAX (Simulación de Entorno) ---")

# 1. INGESTA Y RDD (Simulado con listas de diccionarios)
# Representamos la escalabilidad procesando datos estructurados
# En un entorno real, estos datos podrían provenir de archivos CSV, bases de datos o flujos en tiempo real
datos_ventas = [
    {"Cliente": "C1", "Categoria": "Tecnologia", "Monto": 500.0},
    {"Cliente": "C2", "Categoria": "Ropa", "Monto": 50.0},
    {"Cliente": "C3", "Categoria": "Tecnologia", "Monto": 150.0},
    {"Cliente": "C4", "Categoria": "Hogar", "Monto": 80.0},
    {"Cliente": "C5", "Categoria": "Hogar", "Monto": 200.0}
]

# 2. PROCESAMIENTO ESTRUCTURADO (Equivalente a DataFrames)
# Convertimos la lista de diccionarios a un DataFrame para facilitar el análisis
df_ventas = pd.DataFrame(datos_ventas)
print("\n✅ Paso 1 y 2: Datos cargados en estructura de DataFrame:")
print(df_ventas)

# 3. ANÁLISIS SQL (Equivalente a Spark SQL)
# Simulamos la consulta: SELECT Categoria, SUM(Monto) GROUP BY Categoria
metricas = df_ventas.groupby("Categoria")["Monto"].sum().reset_index()
print("\n✅ Paso 3: Métricas de negocio (Agrupación SQL):")
print(metricas)

# 4. MACHINE LEARNING ESCALABLE (Equivalente a MLlib)
# Usamos KMeans para segmentar clientes por su gasto
X = df_ventas[['Monto']].values
# Definimos 2 grupos: Compradores de bajo volumen y alto volumen
kmeans = KMeans(n_clusters=2, random_state=1, n_init=10)
df_ventas['Segmento_Cluster'] = kmeans.fit_predict(X)

print("\n✅ Paso 4: Pipeline de ML completado (Segmentación K-Means):")
# Renombramos los grupos para que sean legibles
df_ventas['Interpretacion'] = df_ventas['Segmento_Cluster'].map({0: 'Inversión Baja', 1: 'Inversión Alta'})
print(df_ventas[['Cliente', 'Monto', 'Interpretacion']])

print("\n--- ✨ PIPELINE FINALIZADO CON ÉXITO ---")
```

---
### Conclusiones y Notas Técnicas

1. **Análisis de Negocio:** Se identificaron las categorías con mayor volumen de ventas (Tecnología liderando el reporte).
2. **Segmentación:** El modelo K-Means logró diferenciar exitosamente a los clientes entre "Inversión Alta" e "Inversión Baja", permitiendo campañas de marketing dirigidas.
